```python
from pyspark.sql.functions import count, col

dup_df = spark.table("workspace.default.amazon_products_day4_delta")

dup_df.groupBy("asin").agg(count("*").alias("cnt")) \
    .filter(col("cnt") > 1) \
    .orderBy(col("cnt").desc()) \
    .show(20, False)
```

```python
deduped_df = dup_df.dropDuplicates(["asin"])
```

```python
deduped_df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.amazon_products_day4_delta")

print("✅ Duplicates removed and table overwritten.")
```
